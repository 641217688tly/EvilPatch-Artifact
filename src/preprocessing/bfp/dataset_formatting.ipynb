# BFP 原始数据集格式化

对 `data/raw/bfp/` 下的三个原始数据集的 `buggy_code` 和 `fixed_code` 字段做**一次性原地格式化**，
使后续所有依赖这些文件的脚本无需重复格式化。

- `data/raw/bfp/CoCoNut.jsonl`（JSONL 格式）
- `data/raw/bfp/Codeflaws.json`（JSON 数组格式）
- `data/raw/bfp/Deepfix.json`（JSON 数组格式）

**幂等保证**：格式化完成后写入标记文件 `data/raw/bfp/.formatted`，再次运行时自动跳过。

```python
import os
import sys
import json
from pathlib import Path

PROJECT_ROOT = Path(os.path.abspath("")).resolve()
while not (PROJECT_ROOT / "configs").exists() and PROJECT_ROOT != PROJECT_ROOT.parent:
    PROJECT_ROOT = PROJECT_ROOT.parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))
print(f"PROJECT_ROOT: {PROJECT_ROOT}")

from src.utils.formatter import auto_format_c_code_batch

BFP_RAW_DIR   = PROJECT_ROOT / "data/raw/bfp"
FORMATTED_FLAG = BFP_RAW_DIR / ".formatted"

if FORMATTED_FLAG.exists():
    print("[SKIP] 标记文件已存在，BFP 原始数据集已格式化，跳过。")
else:
    print("开始格式化 BFP 原始数据集...")
```

```python
# ========== 格式化 CoCoNut（JSONL 格式）==========
if not FORMATTED_FLAG.exists():
    COCONUT_PATH = BFP_RAW_DIR / "CoCoNut.jsonl"

    records = []
    with open(COCONUT_PATH, encoding="utf-8") as f:
        for line in f:
            line = line.strip()
            if line:
                records.append(json.loads(line))
    print(f"CoCoNut: 读取 {len(records)} 条")

    n = len(records)
    buggy_codes = [r["buggy_code"] for r in records]
    fixed_codes  = [r["fixed_code"]  for r in records]
    all_codes = buggy_codes + fixed_codes

    print(f"  格式化 {len(all_codes)} 条代码文本...")
    results = auto_format_c_code_batch(all_codes, on_error="keep")
    buggy_fmt = [r[0] for r in results[:n]]
    fixed_fmt  = [r[0] for r in results[n:]]

    for record, b, f in zip(records, buggy_fmt, fixed_fmt):
        record["buggy_code"] = b
        record["fixed_code"] = f

    with open(COCONUT_PATH, "w", encoding="utf-8") as f:
        for record in records:
            f.write(json.dumps(record, ensure_ascii=False) + "\n")
    print(f"  ✅ CoCoNut 格式化完成，已原地写回 → {COCONUT_PATH}")
else:
    print("[SKIP] CoCoNut")
```

```python
# ========== 格式化 Codeflaws（JSON 数组格式）==========
if not FORMATTED_FLAG.exists():
    CODEFLAWS_PATH = BFP_RAW_DIR / "Codeflaws.json"

    with open(CODEFLAWS_PATH, encoding="utf-8") as f:
        records = json.load(f)
    print(f"Codeflaws: 读取 {len(records)} 条")

    n = len(records)
    buggy_codes = [r["buggy_code"] for r in records]
    fixed_codes  = [r["fixed_code"]  for r in records]
    all_codes = buggy_codes + fixed_codes

    print(f"  格式化 {len(all_codes)} 条代码文本...")
    results = auto_format_c_code_batch(all_codes, on_error="keep")
    buggy_fmt = [r[0] for r in results[:n]]
    fixed_fmt  = [r[0] for r in results[n:]]

    for record, b, f in zip(records, buggy_fmt, fixed_fmt):
        record["buggy_code"] = b
        record["fixed_code"] = f

    with open(CODEFLAWS_PATH, "w", encoding="utf-8") as f:
        json.dump(records, f, ensure_ascii=False)
    print(f"  ✅ Codeflaws 格式化完成，已原地写回 → {CODEFLAWS_PATH}")
else:
    print("[SKIP] Codeflaws")
```

```python
# ========== 格式化 Deepfix（JSON 数组格式）==========
if not FORMATTED_FLAG.exists():
    DEEPFIX_PATH = BFP_RAW_DIR / "Deepfix.json"

    with open(DEEPFIX_PATH, encoding="utf-8") as f:
        records = json.load(f)
    print(f"Deepfix: 读取 {len(records)} 条")

    n = len(records)
    buggy_codes = [r["buggy_code"] for r in records]
    fixed_codes  = [r["fixed_code"]  for r in records]
    all_codes = buggy_codes + fixed_codes

    print(f"  格式化 {len(all_codes)} 条代码文本...")
    results = auto_format_c_code_batch(all_codes, on_error="keep")
    buggy_fmt = [r[0] for r in results[:n]]
    fixed_fmt  = [r[0] for r in results[n:]]

    for record, b, f in zip(records, buggy_fmt, fixed_fmt):
        record["buggy_code"] = b
        record["fixed_code"] = f

    with open(DEEPFIX_PATH, "w", encoding="utf-8") as f:
        json.dump(records, f, ensure_ascii=False)
    print(f"  ✅ Deepfix 格式化完成，已原地写回 → {DEEPFIX_PATH}")
else:
    print("[SKIP] Deepfix")
```

```python
# ========== 写入完成标记 ==========
if not FORMATTED_FLAG.exists():
    FORMATTED_FLAG.touch()
    print(f"✅ 所有 BFP 原始数据集格式化完成，标记已写入 → {FORMATTED_FLAG}")
else:
    print("[SKIP] 标记文件已存在，无需重复格式化。")
```
